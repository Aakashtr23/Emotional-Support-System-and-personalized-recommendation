```python
import plotly.express as px

# Group by 'score' and count the number of reviews for each score
score_distribution = (df.groupby('score')
                      .agg({'content': 'count'})  # Count the number of 'content' rows
                      .reset_index()
                      .rename(columns={'content': 'count'})  # Rename 'content' to 'count'
                      .sort_values('count', ascending=False))  # Sort by count

# Create the bar plot
fig = px.bar(
    score_distribution,
    x='score',  # Use 'score' as the x-axis
    y='count',  # Use the count of reviews as the y-axis
    color='score',  # Color by score
    opacity=0.7,
    color_discrete_sequence=px.colors.diverging.curl,
    template='plotly_dark',
    title='Label Distribution'
)

# Show the figure
fig.show()
```

```python
def average_word_length(text):
    words = text.split()
    word_count = len(words)
    word_length = 0
    for word in words:
        word_length += len(word)
    avg_word_length = round(word_length / word_count,2)
    return avg_word_length


df['text_length'] = df['content'].apply(lambda x: len(x.split()))
df['average_word_length'] = df['content'].apply(average_word_length)
```

```python
# Viz: Lexical Uniqueness
def unique_word_share(text):
    words = text.split()
    word_count = len(words)
    unique_word_count = len(set([w.lower() for w in words]))
    word_uniqueness = round(unique_word_count * 100/word_count,2)
    return word_uniqueness

df['lexical_uniqueness'] = df['content'].apply(unique_word_share)
```

```python
fig = px.box(df,
             y = 'score',
             x = 'lexical_uniqueness',
             color = 'score',
             color_discrete_sequence = px.colors.diverging.curl,
             template = 'plotly_dark',
             title = 'Boxplot of Lexical Uniqueness per class'
)


fig.show()
```

```python
vec = TfidfVectorizer(max_features=3000)

X_train, X_test, Y_train, Y_test = train_test_split(vec.fit_transform(concatenated_df.content.values).toarray(), 
                                                    concatenated_df.score.values.reshape(-1,1), 
                                                    test_size=0.2, 
                                                    shuffle=True, 
                                                    random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
model = RandomForestClassifier(n_estimators=10,random_state=42)
model.fit(X_train, Y_train)
```

```python
print("Train Accuracy  : {:.2f} %".format(accuracy_score(model.predict(X_train), Y_train)*100))
print("Test Accuracy   : {:.2f} %".format(accuracy_score(model.predict(X_test), Y_test)*100))
print("Precision       : {:.2f} %".format(precision_score(model.predict(X_test), Y_test,average='macro')*100))
print("Recall          : {:.2f} %".format(recall_score(model.predict(X_test), Y_test,average='macro')*100))
```
