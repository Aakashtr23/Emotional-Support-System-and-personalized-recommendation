```python
def choose_random_files(folder_path, num_files=5):
    """
    Choose random files from a specified folder.
    
    Args:
        folder_path (str): Path to the folder containing files.
        num_files (int): Number of random files to select.
    
    Returns:
        list: List of randomly selected file paths.
    """
    print('checking ',folder_path)
    # Get a list of all files in the folder
    files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    
    # Check if the folder contains enough files
    if len(files) == 0:
        raise ValueError("The folder is empty.")
    if num_files > len(files):
        print(f"Requested {num_files} files, but only {len(files)} available. Selecting all files.")
        num_files = len(files)
    
    # Randomly select files
    random_files = random.sample(files, num_files)
    return [os.path.join(folder_path, file) for file in random_files]
```

```python
import os
from IPython.display import Image, display
from PIL import Image as PILImage

# Function to display images from a folder
def display_images_from_folder(folder_path, max_images=10):
    """
    Display images from a specified folder in a Jupyter Notebook.
    
    Args:
        folder_path (str): Path to the folder containing images.
        max_images (int): Maximum number of images to display.
    """
    # Get a list of all files in the folder
    image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))]
    
    # Display each image
    print(f"Displaying up to {max_images} images from: {folder_path}")
    for i, image_file in enumerate(image_files[:max_images]):
        image_path = os.path.join(folder_path, image_file)
        try:
            # Display image
            display(Image(filename=image_path))
            print(f"Image {i+1}: {image_file}")
        except Exception as e:
            print(f"Error displaying {image_file}: {e}")

```

```python
import pygame

def play_mp3(file_path):
    """
    Play an MP3 file using pygame.

    Args:
        file_path (str): Path to the MP3 file to play.
    """
    # Initialize pygame mixer
    pygame.mixer.init()
    try:
        # Load and play the MP3 file
        pygame.mixer.music.load(file_path)
        pygame.mixer.music.play()
        print(f"Playing {file_path}...")

        # Wait until the music finishes playing
        while pygame.mixer.music.get_busy():
            continue
    except pygame.error as e:
        print(f"Error playing file: {e}")

```
